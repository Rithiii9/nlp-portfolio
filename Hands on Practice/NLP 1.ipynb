```python
# Remove stop words and display the top 3 most frequent words.
import nltk
from nltk.corpus import stopwords

text1 = """Natural Language Processing is a branch of artificial intelligence that helps
computers understand, interpret, and generate human language. It is used in
applications such as chatbots, machine translation, sentiment analysis, and
speech recognition. Learning NLP provides opportunities to build intelligent
systems that can process large amounts of text efficiently."""

stop_words = set(stopwords.words('english'))
words = [word for word in text1.split() if word.lower() not in stop_words]
fd = nltk.FreqDist(words)
print(fd.most_common(3))
```

```python
# Group words by their length using Conditional Frequency Distribution (CFD).
from nltk.probability import ConditionalFreqDist
from nltk.corpus import inaugural
words = inaugural.words("1861-Lincoln.txt")
cfd = ConditionalFreqDist((len(w), w) for w in words)
cfd[6]
```

```python
# Find the top 5 presidents who used the longest words(sounding words) in their inaugural speeches using CFD.
from nltk.corpus import inaugural
from nltk.probability import ConditionalFreqDist

cfd = ConditionalFreqDist(
    (fileid, word)
    for fileid in inaugural.fileids()
    for word in inaugural.words(fileid)
)

result = []

for fileid in inaugural.fileids():
    longest = max(cfd[fileid], key=len)
    result.append((fileid, longest, len(longest)))

result.sort(key=lambda x: x[2], reverse=True)

for fileid, word, length in result[:5]:
    print(fileid, ":", word, "-", length)
```
