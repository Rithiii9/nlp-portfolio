```python
def gender_features(word):
    return {'last_letter': word[-1]}

gender_features('Obama')
```

```python
from nltk.corpus import names
names.words()
print(len(names.words()))
#print(names.words())

labeled_names = ([(name,'male') for name in names.words('male.txt')] + [(name, 'female')  for name in names.words('female.txt')])
print(labeled_names[:5])

import random
random.shuffle(labeled_names)

featuresets = [(gender_features(n), gender) for (n,gender) in labeled_names]
print(featuresets[:5])

#Try Printing featurests

train_set, test_set = featuresets[5000:], featuresets[:2000]

import nltk
classifier = nltk.NaiveBayesClassifier.train(train_set)

print(classifier.classify(gender_features('Obama')))
print(classifier.classify(gender_features('Yann Lecun')))
print(classifier.classify(gender_features('Vinytha')))
print(classifier.classify(gender_features('David')))


print(nltk.classify.accuracy(classifier,test_set))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vect = CountVectorizer(binary =True) # Explore other parameters too!!
corpus = ["Tessaract is good optical character recognition engine  ", "optical character recognition is significant "]

vect.fit(corpus) #learns the unique words (Vocab) in the corpus

#vect.vocabulary_ #stores the vocab  #vocabulary_  built-in:-)
vocab = vect.vocabulary_

for key in sorted(vocab.keys()):
    print("{}:{}".format(key, vocab[key]))

print(vect.transform(["This is a good optical illusion"]).toarray())

#Finding Similarity Between the Documents
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vect.transform(["Google Cloud Vision is a character recognition engine"]).toarray(), vect.transform(["OCR is an optical character recognition engine"]).toarray())
print(similarity)
```
